```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('../data/ev_market_2026.csv')
df.head()
```

```python
brand_count = df.groupby('brand').count().reset_index()
plt.barh(brand_count['brand'], brand_count['model'])
plt.title('Data Count for Each Brand')
plt.ylabel('Brand')
plt.xlabel('Count')
plt.gca().invert_yaxis()  
plt.show()
```

Skewed data, brand with more data entry would be more accurate (law of large numbers)

```python
models_per_brand = df[['brand', 'model']].drop_duplicates('model').groupby('brand').count().reset_index()
plt.barh(models_per_brand['brand'], models_per_brand['model'])
plt.title('Number of Models per Brand')
plt.ylabel('Brand')
plt.xlabel('Number of unique models')
plt.gca().invert_yaxis()  
plt.show()
```

```python
price = df['price_usd']
plt.hist(price)
plt.title('Price Distribution')
plt.xlabel('Price (USD)')
plt.ylabel('count')
plt.show()
```
